```python
# Celda 3: Código principal del bot Telegram integrando bot_utils

# IMPORTANT: Replace "TU_TOKEN_TELEGRAM" with your actual Telegram bot token.
# You can get a token from BotFather on Telegram.
TELEGRAM_TOKEN = "8509265501:AAEMAHVgn9s2SQOGKOXp67izXsk_91F1Cws"  # Pon tu token aquí
bot = telebot.TeleBot(TELEGRAM_TOKEN)
cookie_actual = None
email_global = None
password_global = None
faker = Faker()


def obtener_cookies_con_selenium(email, password):
    options = Options()
    options.add_argument("--headless")
    options.add_argument("--no-sandbox")
    options.add_argument("--disable-dev-shm-usage")
    options.add_argument("--remote-debugging-port=9222")
    options.add_argument('--disable-gpu')
    options.add_argument('--window-size=1920,1080')

    driver = webdriver.Chrome(options=options)
    driver.get("https://www.amazon.com.mx/ap/signin")

    input_email = driver.find_element(By.ID, "ap_email")
    input_email.send_keys(email)
    driver.find_element(By.ID, "continue").click()
    time.sleep(2)

    input_password = driver.find_element(By.ID, "ap_password")
    input_password.send_keys(password)
    driver.find_element(By.ID, "signInSubmit").click()
    time.sleep(5)

    cookies = driver.get_cookies()
    cookie_string = "; ".join([f"{c['name']}={c['value']}" for c in cookies])

    driver.quit()
    return cookie_string


@bot.message_handler(commands=['start'])
def cmd_start(message):
    bot.send_message(message.chat.id,
                     "Hola! Usa /login para iniciar sesión en Amazon.\nUsa /check numero|mm|aaaa|cvv para validar tarjetas.\nUsa /cookies para obtener las cookies actuales.")


@bot.message_handler(commands=['login'])
def cmd_login(message):
    chat_id = message.chat.id
    try:
        msg = bot.send_message(chat_id, "Envía tu email de Amazon:")
        bot.register_next_step_handler(msg, process_email)
    except Exception as e:
        bot.send_message(chat_id, f"Error: {str(e)}")


def process_email(message):
    global email_global
    email_global = message.text.strip()
    chat_id = message.chat.id
    msg = bot.send_message(chat_id, "Ahora envía tu contraseña:")
    bot.register_next_step_handler(msg, process_password)


def process_password(message):
    global password_global, cookie_actual, email_global
    password_global = message.text.strip()
    chat_id = message.chat.id
    bot.send_message(chat_id, "Iniciando sesión y obteniendo cookies, espera...")
    try:
        cookie_actual = obtener_cookies_con_selenium(email_global, password_global)
        bot.send_message(chat_id, "¡Login exitoso! Cookies guardadas.")
    except Exception as e:
        bot.send_message(chat_id, f"Error durante login: {str(e)}")


@bot.message_handler(commands=['cookies'])
def cmd_cookies(message):
    global cookie_actual, email_global, password_global
    chat_id = message.chat.id
    if not email_global or not password_global:
        bot.send_message(chat_id, "No hay credenciales guardadas. Usa /login primero.")
        return
    bot.send_message(chat_id, "Re-obteniendo cookies con las credenciales guardadas, espera...")
    try:
        cookie_actual = obtener_cookies_con_selenium(email_global, password_global)
        bot.send_message(chat_id, f"Cookies actualizadas:\n{cookie_actual}")
    except Exception as e:
        bot.send_message(chat_id, f"Error al obtener cookies: {str(e)}")


@bot.message_handler(commands=['check'])
def cmd_check(message):
    global cookie_actual
    chat_id = message.chat.id
    if not cookie_actual:
        bot.send_message(chat_id, "No hay cookies disponibles. Usa /login para iniciar sesión primero.")
        return
    tarjeta = message.text.replace('/check', '').strip()
    if not tarjeta:
        bot.send_message(chat_id,
                         "Por favor envía la tarjeta con el comando /check seguido del número en formato numero|mm|aaaa|cvv")
        return
    gate = bot_utils.AmazonMXGate(card=tarjeta, cookie=cookie_actual)
    resultado = gate.check()
    estado = "✅ APROBADA" if resultado.get('success') else "❌ DECLINADA"
    bot.send_message(chat_id, f"{estado}\nDetalles: {resultado.get('response')}")


print("Bot iniciado, esperando mensajes...")
bot.infinity_polling()
```
